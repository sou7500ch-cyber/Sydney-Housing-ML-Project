```python
import os

DATA_FOLDER = "."

dat_files = []

for root, folders, files in os.walk(DATA_FOLDER):
    for file in files:
        if file.lower().endswith(".dat"):
            dat_files.append(os.path.join(root, file))

print("Number of DAT files:", len(dat_files))
```

```python
target_suburbs = {
    "CHATSWOOD",
    "PARRAMATTA",
    "BLACKTOWN"
}

records = []

for filepath in dat_files:

    with open(filepath, "r", encoding="latin1", errors="replace") as file:

        for line in file:

            if not line.startswith("B;"):
                continue

            fields = line.rstrip("\n\r").split(";")

            if len(fields) < 24:
                continue

            suburb = fields[9].strip().upper()

            if suburb not in target_suburbs:
                continue

            records.append({
                "Property_ID_Raw": fields[2].strip(),
                "Sale_Counter": fields[3].strip(),
                "Street_Number": fields[7].strip(),
                "Street_Name": fields[8].strip(),
                "Suburb": suburb.title(),
                "Postcode": fields[10].strip(),
                "Land Size": fields[11].strip(),
                "Contract_Date": fields[13].strip(),
                "Settlement_Date": fields[14].strip(),
                "Sale_Price": fields[15].strip(),
                "Zoning": fields[16].strip(),
                "Property_type": fields[18].strip(),
                "Dealing_Number": fields[23].strip()
            })

df = pd.DataFrame(records)

print("Total extracted records:", len(df))
df.head(10)
```

```python
# Create the initial modelling dataset

model_columns = [
    "Suburb",
    "Land Size",
    "Contract_Date",
    "Zoning",
    "Sale_Price"
]

df_ml = df_model[model_columns].copy()

print("ML dataset shape:", df_ml.shape)
print("\nColumns:")
print(df_ml.columns.tolist())
```

```python
# Convert Contract_Date to datetime
df_ml["Contract_Date"] = pd.to_datetime(
    df_ml["Contract_Date"],
    errors="coerce"
)

# Extract useful date features
df_ml["Contract_Year"] = df_ml["Contract_Date"].dt.year
df_ml["Contract_Month"] = df_ml["Contract_Date"].dt.month

# Check the result
print(df_ml[[
    "Contract_Date",
    "Contract_Year",
    "Contract_Month"
]].head(10))
```

```python
print(df_ml["Contract_Year"].value_counts().sort_index())
```

```python
df_ml = df_ml.drop(columns=["Contract_Date"])

print(df_ml.head())
print("\nShape:", df_ml.shape)
```

```python
# Check missing values in the ML dataset

print(df_ml.isnull().sum())
```

```python
print("\nData types:")
print(df_ml.dtypes)
```

```python
from sklearn.preprocessing import OneHotEncoder

# Create encoder
encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

# Encode categorical columns
encoded = encoder.fit_transform(
    df_ml[["Suburb", "Zoning"]]
)

# Get encoded column names
encoded_columns = encoder.get_feature_names_out(
    ["Suburb", "Zoning"]
)

# Create DataFrame
df_encoded = pd.DataFrame(
    encoded,
    columns=encoded_columns,
    index=df_ml.index
)

# Combine numerical columns with encoded columns
df_final = pd.concat(
    [
        df_ml[[
            "Land Size",
            "Contract_Year",
            "Contract_Month",
            "Sale_Price"
        ]],
        df_encoded
    ],
    axis=1
)

print("Final dataset shape:", df_final.shape)
print("\nColumns:")
print(df_final.columns.tolist())
print(df_final.head())
```

```python
# Separate features and target

X = df_final.drop(columns=["Sale_Price"])
y = df_final["Sale_Price"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training X shape:", X_train.shape)
print("Testing X shape:", X_test.shape)
print("Training y shape:", y_train.shape)
print("Testing y shape:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create the model
linear_model = LinearRegression()

# Train the model
linear_model.fit(X_train, y_train)

# Make predictions
y_pred = linear_model.predict(X_test)

# Evaluate
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 4))
```
